```python
import os
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator

DATASET_PATH = 'PlantVillage-Dataset-master/raw/color'
IMAGE_SIZE = (224, 224)
BATCH_SIZE = 32

# Set up the paper's realistic data augmentation settings
train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=20,
    width_shift_range=0.2,
    height_shift_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True,
    vertical_flip=True,
    validation_split=0.2 # Reserves 20% of data for validation monitoring
)

# Load real training images (80%)
train_generator = train_datagen.flow_from_directory(
    DATASET_PATH,
    target_size=IMAGE_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    subset='training'
)

# Load real validation images (20%)
val_generator = train_datagen.flow_from_directory(
    DATASET_PATH,
    target_size=IMAGE_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    subset='validation'
)

# Dynamically get the exact number of plant classes found in the dataset folder
NUM_CLASSES = train_generator.num_classes
print(f"Successfully loaded {NUM_CLASSES} unique plant disease classes!")
```

```python
import time
from tensorflow.keras import layers, models, optimizers

# Build your baseline ResNet50 architecture
base_model = tf.keras.applications.ResNet50(
    input_shape=(224, 224, 3),
    include_top=False,
    weights='imagenet'
)
base_model.trainable = True

model = models.Sequential([
    base_model,
    layers.GlobalAveragePooling2D(),
    layers.Dense(128, activation='relu'),
    layers.Dropout(0.5),
    layers.Dense(NUM_CLASSES, activation='softmax')
])

# Compile the neural network
model.compile(
    optimizer=optimizers.Adam(learning_rate=0.001),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# Start training on the real dataset
print("\nInitiating Model Training Loop on GPU...")
start_time = time.time()

history = model.fit(
    train_generator,
    epochs=5, # Let's start with 5 epochs to verify completion speed
    validation_data=val_generator,
    verbose=1
)

execution_duration = time.time() - start_time
print(f"Training Complete! Total Run Duration: {execution_duration:.2f} seconds")

# Save the trained model checkpoint
model.save("plant_disease_baseline_model.h5")
```

```python
import matplotlib.pyplot as plt

# Generate performance verification charts
plt.figure(figsize=(12, 5))

# Chart 1: Categorical Loss Trajectories
plt.subplot(1, 2, 1)
plt.plot(history.history['loss'], label='Our Training Loss', color='#1f77b4', linewidth=2)
plt.plot(history.history['val_loss'], label='Our Validation Loss', color='#ff7f0e', linewidth=2, linestyle='--')
plt.title('Baseline ResNet50: Loss Trajectory', fontsize=12, fontweight='bold')
plt.xlabel('Epochs')
plt.ylabel('Loss Value')
plt.legend()
plt.grid(True, alpha=0.3)

# Chart 2: Accuracy Optimization Curve
plt.subplot(1, 2, 2)
plt.plot(history.history['accuracy'], label='Our Training Accuracy', color='#2ca02c', linewidth=2)
plt.plot(history.history['val_accuracy'], label='Our Validation Accuracy', color='#d62728', linewidth=2, linestyle='--')
plt.title('Baseline ResNet50: Accuracy Growth', fontsize=12, fontweight='bold')
plt.xlabel('Epochs')
plt.ylabel('Accuracy Score')
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('baseline_evaluation_curves.png', dpi=300)
plt.show()
print("Graphs saved locally as 'baseline_evaluation_curves.png'")
```
